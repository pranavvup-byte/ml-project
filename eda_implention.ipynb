```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/sample_data/CKD_NHANES_2021_2023.csv')

print("Dataset loaded successfully!")
```

```python
print("First 5 rows:")
display(df.head())
```

```python
print("Last 5 rows:")
display(df.tail())
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Column Names:")
print(df.columns.tolist())
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
missing = df.isnull().sum()

print("Columns with missing values:")
print(missing[missing > 0])
```

```python
print("Number of duplicate rows:", df.duplicated().sum())
```

```python
print("Statistical Summary:")
display(df.describe())
```

```python
plt.figure(figsize=(6, 5))

df['ckd_present'].value_counts().plot(kind='bar')

plt.title('CKD Target Distribution')
plt.xlabel('CKD Present')
plt.ylabel('Number of Records')
plt.xticks(rotation=0)

plt.show()
```

```python
numeric_columns = df.select_dtypes(
    include=['int64', 'float64', 'int32', 'float32']
).columns

print("Numerical Columns:")
print(numeric_columns.tolist())
```

```python
df[numeric_columns].hist(
    figsize=(15, 12),
    bins=20
)

plt.suptitle("Distribution of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

df[numeric_columns].boxplot()

plt.title("Boxplot of Numerical Features")
plt.xticks(rotation=90)
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df['age'], bins=20, kde=True)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='ckd_present',
    y='age',
    data=df
)

plt.title("Age vs CKD Status")
plt.xlabel("CKD Present")
plt.ylabel("Age")

plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = df.select_dtypes(
    include=np.number
).corr()

sns.heatmap(
    correlation,
    annot=False,
    cmap='coolwarm'
)

plt.title("Correlation Matrix")

plt.show()
```

```python
print("Potential leakage / identifier columns:")

for col in ['participant_id', 'ckd_stage', 'eGFR']:
    if col in df.columns:
        print(col, "-> present in dataset")
```
